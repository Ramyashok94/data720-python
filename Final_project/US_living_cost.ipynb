```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
us_livingCost_data = pd.read_csv("cost_of_living_us.csv")
us_livingCost_data.head()
```

```python
us_livingCost_data.isna().sum()
```

```python
#since median_family_income has 10 null value , we are dropping those rows.

us_livingCost_data= us_livingCost_data.dropna(subset = ["median_family_income"])
us_livingCost_data.tail()
```

```python
#validating missing data after removing the null rows
us_livingCost_data.isna().sum()
```

```python
us_livingCost_data.info()
```

```python
us_livingCost_data.describe()
```

```python
us_livingCost_data.set_index("state")
```

```python
#Add new columns for parent count and children count

parent_count = us_livingCost_data["family_member_count"].str.extract(r'(\d+)p').astype(int) #take the integer value before letter 'p'
children_count = us_livingCost_data["family_member_count"].str.extract(r'p(\d+)c').astype(int) # take the integer value between p and c
#Inserting to the dataframe

us_livingCost_data.insert(5,"parent_count",parent_count)
us_livingCost_data.insert(6,"children_count",children_count)
us_livingCost_data.head()
```

```python
#Visualizing the total cost for each state to find the outliers
plt.figure(figsize=(20,10))
sns.scatterplot(us_livingCost_data, x= "state", y="total_cost")
plt.xticks(rotation=90)
plt.show()
#Note: DC has less family expense collected than other states
```

```python


#Calculating the average expenditure for each state.

mean_cost = us_livingCost_data.groupby("state")[["total_cost"]].mean()
avg_total_cost = mean_cost["total_cost"]
state_count = us_livingCost_data["state"].value_counts()

avg_cost = (avg_total_cost*state_count/state_count.sum())
avg_cost.head()
```

```python

#ploting the average cost for each state
plt.figure(figsize=(10,6))
sns.barplot(avg_cost)
plt.ylabel("Average total cost")
plt.xlabel("States in US")
plt.xticks(rotation=90)
plt.title("Comparison of Average living cost for the states in US")
plt.show()

#Avg living cost of family in exas in high and DC has less 
```
