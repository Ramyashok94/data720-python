```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the airline losses data
df = pd.read_csv("Aviation_disruption/airline_losses_estimate.csv")

# Prepare KPI values
total_loss = df["estimated_daily_loss_usd"].sum() / 1e9  # Convert to billions
airlines_affected = df["airline"].nunique()
total_cancellations = df["cancelled_flights"].sum()
avg_revenue_loss = df["estimated_daily_loss_usd"].mean() / df["estimated_daily_loss_usd"].sum() * 100

kpis = [
    ('Total loss',          f"${total_loss:.1f}B"),
    ('Airlines affected',   f"{airlines_affected}"),
    ('Total cancellations', f"{total_cancellations:,}"),
    ('Avg revenue loss',    f"{avg_revenue_loss:.1f}%"),
]
ACCENT = '#D85A30'

fig, axes = plt.subplots(1, 4, figsize=(16, 3))
fig.patch.set_facecolor('#0f1117')

for ax, (label, value) in zip(axes, kpis):
    ax.set_facecolor('#1e2235')
    for spine in ax.spines.values(): spine.set_visible(False)
    ax.set_xticks([]); ax.set_yticks([])
    ax.text(0.5, 0.62, value, ha='center', va='center',
            fontsize=28, fontweight='bold', color=ACCENT, transform=ax.transAxes)
    ax.text(0.5, 0.22, label, ha='center', va='center',
            fontsize=12, color='#8a8d9a', transform=ax.transAxes)

plt.suptitle('Airline Losses — Key Metrics', y=1.04,
             fontsize=15, fontweight='bold', color='white')
plt.tight_layout()
plt.show()
```
