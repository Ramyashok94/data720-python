```python
# Write the code for your program below.
import re

# Remove punctuation and convert all words to lowercase.
def cleanLines(words):
    removePunctuation = [re.sub('[^\w\s]|\n','',word.lower()) for word in words]
    return removePunctuation
```

```python
''' Use a set to find all unique words.
    Use a dictionary to count the frequency of each unique word.'''

def uniqueWordsAndCount(inputList):
    #creating a empty set
    uniqueSet = set()
    #creating an empty dictionary
    countDict = {}
    for value in inputList:
        '''Adding vlaues to the set'''
        uniqueSet.add(value)
        # checking the occurance of words
        if value in countDict:
            countDict[value] += 1
        else:
            countDict[value] = 1
        
    return countDict
```

```python
def readAndOrganizefile(uploadedFile):
    # Read the content of the text file `Unit2_Python_learning_journey.txt`.
    file = open(uploadedFile)
    lineStr = file.read()

    # Use a set to find all unique words.
    wordsList = lineStr.split()
    #calling cleanLines()
    cleanLinelist =cleanLines(wordsList) 

  
   # calling uniqueWordsAndCount()
    uniqWordCount = uniqueWordsAndCount(cleanLinelist)
    return [(word,count) for word,count in uniqWordCount.items()]
```
