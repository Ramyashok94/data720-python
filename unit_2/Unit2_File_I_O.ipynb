## Unit 2 File I/O Quiz
* The objective of this quiz is for you to write code that reads data from a file and does simple analysis on the data.
* The name of the file is `Unit2_city_weather_data.csv`.
* The file contains weather data for two North Carolina cities.
* The file is in the CSV format.
* The file has three columns: a date, a city and the temperature for that city and date.
* Your code should do the following:
    * Read the data from Unit2_city_weather_data.txt.
    * Extract the date, city, and temperature values.
    * Calculate the average temperature for each city.
    * Print out the average temperature for each city.
    * Identify and print the dates for each city where the temperature was above the city's average.
* **Note: Do not use Pandas to read in data**

```python
# Write the code for your program below.
import csv

date =[]
city =[]
temperatureStr=[]
temperature = []
with open('Unit2_city_weather_data.csv', mode ='r') as newFile:

    ''' Read the data from Unit2_city_weather_data.txt. '''
    
    read = csv.reader(newFile)
    # to skip the title of the column
    next(read) 
    
    ''' Extract the date, city, and temperature values. '''
    
    for lines in read:
        date.append(lines[0])
        city.append(lines[1])
        temperatureStr.append(lines[2])
```

```python
# changing the elements to int
for temp in temperatureStr:
    temp = int(temp)
    temperature.append(temp)

listOfCityWeather =list(zip(date,city,temperature))
```

```python
''' Calculate the average temperature for each city. '''

raleighTemp = []
charlotteTemp = []
for date,city,temp in listOfCityWeather:
    if city == 'Raleigh' :
        raleighTemp.append(temp)
    else:
        charlotteTemp.append(temp)   

raleightCount = len(raleighTemp)
raleighTempTotal = sum(raleighTemp)
raleighTempMean = raleighTempTotal/raleightCount

charlotteCount = len(charlotteTemp)
charlotteTempTotal= sum(charlotteTemp)
charlotteTempMean = charlotteTempTotal/charlotteCount

```

```python
''' Print out the average temperature for each city. '''

print("Raleigh average Temperature: ",raleighTempMean)
print("Charlotte average Temperature: ",charlotteTempMean)

''' Identify and print the dates for each city where the temperature was above the city's average.'''

print("Below are the dates and cities when the temperature is above average: ")
for date,city,temp in listOfCityWeather:
    if city == 'Raleigh' and  temp > raleighTempMean :
        print(f'On {date} at {city}')
    if city == 'Charlotte' and  temp > charlotteTempMean :
        print(f'On {date} at {city}')
```
