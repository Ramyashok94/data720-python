```python
# Write the code for your program below.
import random

def guess_random_num(numRange,no_of_guess):
    randomNum = random.randint(1,numRange) 
    print('''Game rules:
    
Guess the random number between 1 and 100.

If your guess matches the random number, you win.

You have 6 turns to guess the correct number.

After each guess, you may receive hints (like “too big” or “too small”) to guide your next guess.

If you don’t guess correctly within 6 turns, the game ends and the number is revealed.

Get started!''') 
    print('-'*100) 
    for i in range(1,no_of_guess+1): 
        player_chose = int(input(f'Enter Guess {i} :').strip())
        if player_chose == randomNum:
            print(f'Random number is {randomNum}. \nCongratulations!!! you won!!👏🎉''') 
            break
        elif i < no_of_guess and player_chose > randomNum :
            print(f'Too big! Try a smaller number.')
        elif i < no_of_guess and player_chose < randomNum :
            print(f'Too small! Try a bigger number.')
            
    else: 
        print(f'Random number is {randomNum}. \nSorry,you lost!!!😔your turns are over. Thanks for participanting') 


guess_random_num(100,6)
```
