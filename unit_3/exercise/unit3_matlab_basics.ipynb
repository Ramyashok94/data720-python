```python
x = pv.pops
y = pv.total
plt.scatter(x,y)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x,y)
plt.xscale("log") # logs of make distance 1 10 100 100 , equal ratio 
plt.yscale("log")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x,y, s  = pv.coal , c=pv.per_capita , cmap = "viridis")
# s-> size (it takes the variable value as size)
# c -> color (it takes the variable value as color)
# cmap -> gives the scale for color map
#alpha -> gives opacity /transparent color 
plt.xscale("log") # logs of make distance 1 10 100 100 , equal ratio 
plt.yscale("log")
plt.xlabel("population")
plt.colorbar(label = "per capita emission")
plt.ylabel("co2 emission")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x,y, s  = pv.coal , c=pv.per_capita , cmap = "viridis",alpha = 0.6
           )
# s-> size (it takes the variable value as size)
# c -> color (it takes the variable value as color)
# cmap -> gives the scale for color map
#alpha -> gives opacity /transparent color 
plt.xscale("log") # logs of make distance 1 10 100 100 , equal ratio 
plt.yscale("log")
plt.xlabel("population")
plt.colorbar(label = "per capita emission")
plt.ylabel("co2 emission")
plt.xlim([10**4 , 10**10])
plt.ylim([10**-1, 10**5])
plt.show()
```
