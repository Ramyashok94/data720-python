```python
 x = np.linspace(1,10,100)
y_sin = np.sin(x)
y_cos = np.cos(x)

# subplot for 2 rows and one column
fig,axes = plt.subplots(nrows=2, ncols=1)

# ploting using axes since plt.plot takes only the latest
axes[0] .plot(x,y_sin)
axes[0] .set_title("sin x")

# ploting using axes since plt.plot takes only the latest
axes[1].plot(x,y_cos)
axes[1].set_title("cos x")

plt.show()
```

Bar chart for categorical dataset

```python
x= pv. country
y = pv.total

plt.figure(figsize = (8,6))
plt.bar(x,y)
plt.xlabel("country")
plt.ylabel("total")
plt.show()
```

```python
high_value = [i for i,total in enumerate(pv.total) if total>100]
x= [pv.country[i]  for i in high_value]
y = [pv.total[i] for i in high_value]

plt.figure(figsize = (8,6))
plt.bar(x,y)
plt.xlabel("Country")
plt.ylabel("Total")
plt.xticks(rotation=90)
plt.show()
```

```python
color =[]
for total in y:
    if total>1000:
        color.append("red")
    else:
        color.append("blue")
    
```

```python
plt.figure(figsize = (8,6))
plt.bar(x,y,color = color)
plt.xlabel("Country")
plt.ylabel("Total")
plt.xticks(rotation=90)
plt.show()
```

```python
label_dict ={
    "USA" : 0,
    "China" : 0,
    "India" :0,
    "Everyone else" : 0
}

for country,total in zip(pv.country, pv.total):
    if country in label_dict:
        label_dict[country] +=total
    else:
        label_dict["Everyone else"]+=total

print(label_dict)
labels = list(label_dict.keys())
slices= list(label_dict.values())

plt.pie(slices, labels=labels) # labels(keyword in syntax) = value because plt.pie(x, labels=None, autopct=None, colors=None, startangle=None)
plt.show()
    
```
