#<b>Bank Account:
<p>To prevent the variable from accessing we use "__"

```python
class BankAccount:
    def __init__(self, account_num, balance):
        self.__account_num = account_num
        self.__balance= balance

    def deposit(self,amount):
         self.__balance += amount

    def withdrawal(self,amount):
         self.__balance -= amount
    def get_balance(self):
        return self.__balance
```

```python
account = BankAccount(1234,1000)
```

```python
account.deposit(1000)
```

```python
account.get_balance()
```

```python
account.withdrawal(500)
```

```python
account.get_balance()
```

```python
account.__balance
```

#<b>Magic Functions

```python
class Vehicle:
    def __init__(self, mpg=None, tank_size=None,passenger=None, transmission=None):
        self.mpg = mpg
        self.tank_size= tank_size
        self.passenger = passenger
        self.transmission = transmission
        self.max_range = self.calculate_max_range()
        self.rating = self.rating

    def calculate_max_range(self):
        return self.mpg*self.tank_size

    #Magic function for the below print function

    def __str__(self):
        string = f'''
            fuel efficiency = {self.mpg}
            tank size = {self.tank_size}
            passenger = {self.passenger}
            transmission = {self.transmission}
            '''
        return string

    def __repr__(self):
         string1 = f'''
            fuel efficiency = self.mpg
            tank size = {self.tank_size}
            passenger = {self.passenger}
            transmission = {self.transmission}
            '''
         return string1


    #Not equal magic function

    def __ne__(self,other):
        if isinstance(other,Vehicle):
            return self.rating != other.rating
        else:
            return False

    def __gt__(self,other):
        if instance(other,Vehicle):
            return self.rating > other.rating
        else:
            return False
            

def print_vehicle(vehicle):
    print(f'''
    fuel efficiency = {vehicle.mpg}
    tank size = {vehicle.tank_size}
    passenger = {vehicle.passenger}
    transmission = {vehicle.transmission}
    ''')
```

```python
sedan = Vehicle(mpg=35,tank_size=15,passenger=5,transmission="auto")
print_vehicle(sedan)
```

```python
class ElectricVehicle(Vehicle):
    def __init__(self, battery_capacity, battery_efficiency,passenger,rating):
        self.battery_capacity = battery_capacity # in kWh
        self.battery_efficiency = battery_efficiency # in kWh/mile
        super().__init__(passenger=passenger,rating=rating)
    
```

```python
testla = ElectricVehicle(battery_capacity=34,battery_efficiency= 0.56,passenger=5,rating=3)
volvo = ElectricVehicle(battery_capacity=45,battery_efficiency= 0.66,passenger=5,rating=4)
print(testla)
testla.rating>volvo.rating
```
