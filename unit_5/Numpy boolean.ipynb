```python
import numpy as np
A=  np.array([3,4,6,10,24,89,45,43,46,99,100])
#which are not divisible by 3
B = A[(A%3 != 0)]
print("which are not divisible by 3: \n", B)
```

```python
#which are divisible by 5

c =A[(A%5 == 0)] 
print("which are divisible by 5: \n", c)
```

```python
# which are divisible 3 and 5
D = A[(A%3==0)&(A%5==0)]
print("which are divisible by 3 and 5: \n", D)
```

```python
#which are divisible by 3 and set them to 42

A[A%3==0] = 42
print("which are divisible by 3 and set them to 42: \n",A)
```

```python
import numpy as np

a = np.array([[0, 2, 3, 0, 1],
              [1, 0, 0, 7, 0],
              [5, 0, 0, 1, 0]])

print(a.nonzero())
print(np.transpose(a.nonzero()))

#non-zero values can be retrieved using below code
print(a[a.nonzero()])
#flatnonzero
print(np.flatnonzero(a))
print(np.count_nonzero(a))
```
