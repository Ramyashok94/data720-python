#### Numpy arrays are all same datatype.

All elements in an array of numpy must be same data type.
Assume if the first element arr[0] is an integer than the ellemts in the array will hold only integer value.
Where `list` in python are `heterogenous` which holds different datatype, the `array` in numpy are `homogeneous` data types.
Lets see an example below trying to insert an float value in an integer array of numpy.

```python
## creating an array using numpy

arr = np.arange(15)
print(arr)

## checking the length of array using len() function
print(len(arr))

## inserting a float value in the middle of an integer array

arr[(len(arr) // 2)] = 36.5
print(arr)
```

In the above resulting array , you can notice that the `float value` is inserted as integer since the numpy array holds same type, it converts the `float` to `int` and inserted to the index specified.

Since float is a numeric value it converted it to Integer.
whatif we assign a value that cant be converted in integer?

```python
arr[4]="arr"
print(arr)
```

The above trying to insert the string value , which cant be converted so throwing  `ValueError`.
Since it throwing error enclosing it in the `try except` for smooth running of the program.

```python
import sys
try:
    arr[4]="arr"
    print(arr)
except:
    print("Exception :", sys.exc_info()[1])
    pass
```

#### Numpy has zeros() to have zero array
##### Syntax:
`np.zeros(num,dtype=float)` 

it takes two arg first takes number of elements, second parameter is `dtype` , which is by default assign to `float`. If we want we can change into int or bool `dtype=int`,`dtype=bool` . we can see all examples below.


```python
## this gives float value.
zeroArr = np.zeros(10)
print("float value array:",zeroArr)

## changing to int type
zeroArrInt = np.zeros(10,dtype=int)
print("Integer value array:",zeroArrInt)

## changing to bool data type
zeroArrBool = np.zeros(5,dtype=bool)
print("Boolean value array:",zeroArrBool)

## Creating another bool array in different way
anotherBool = np.array([False]*5)
print("Boolean value array:",anotherBool)


```

#### np.any and np.all
-`np.any` - evaluates if any one element in both array matches   
-`np.all` - evaluates wether all the elements in both the arrays matches  
-syntax:       
`np.any(first array `==` second array)`        
`np.all(first array `==` second array)`

```python
any_element_matches = np.any(zeroArrBool == anotherBool)
all_element_matches = np.all(zeroArrBool == anotherBool)
print("Any elements in two array matched: ", any_element_matches)
print("All elements in two array matched: ", all_element_matches)
```
