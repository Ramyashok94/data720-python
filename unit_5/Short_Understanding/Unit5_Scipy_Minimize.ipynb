```python
# Import necessary libraries
import numpy as np
from scipy.optimize import minimize
import matplotlib.pyplot as plt
```

```python
#function

def f(x):
    ''' computes f(x) = x^2 + 10 * sin(x)
    parameter x : input value
    return f(x)'''
    return x**2 + 10 * np.sin(x)
```

```python
# fake data 
'''Created 50 evenly spaced x value from 0 to 2'''
x = np.linspace(0,2,50)
'''calls the function'''
y=f(x)

'''Creates a figure'''
plt.figure(figsize=(8,5))
'''plots x and y on the graph'''
plt.plot(x,y,lw=2)
plt.grid()
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = x^2 + 10 * sin(x)")
plt.show()
```
