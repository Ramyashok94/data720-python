```python

import comp116
import pickle

with open("Unit-5-4-Numpy.data.pickle",'rb') as file:
   (state_names, state_co2, co2_categories)= pickle.load(file)

   print(type(state_co2))
   print(type(state_names))
   print(type(co2_categories))
```

```python
comp116.array_to_html(state_co2, row_names=state_names,col_names=co2_categories)
```

```python
# Design pattern for extracting a row from the file

#finding the idaho in the statename
print(state_names == "Idaho")
Idaho_array = state_names == "Idaho"
Idaho_offset = np.argmax(Idaho_array)
#print(Idaho_offset)
```

```python
# extracting the Idaho details from the table
Idaho_details = state_co2[Idaho_offset,:]
print(Idaho_details)
# inorder to avoid the length issue since comp116 accepts only 2D array
print(Idaho_details.shape)
comp116.array_to_html(np.reshape(Idaho_details,(1,len(Idaho_details))),
                      row_names=["Idaho"],col_names=co2_categories)
```

```python
# Design pattern for extracting a column from the file
print("co2_categories in the dataset: ",co2_categories)

nat_gas_array = co2_categories =="Transportation nat gas"
print("Array of nat gas: ",nat_gas_array)

nat_gas_offset = np.argmax(nat_gas_array)
print("Offset of nat gas: ",nat_gas_offset)

nat_gas = state_co2[:,nat_gas_offset]
print(nat_gas)

comp116.array_to_html(np.reshape(nat_gas,(len(nat_gas),1)),row_names=state_names,col_names=["Transportation nat gas"])
```

```python
#multiple rows using boolean rows
#florida or kanasa row selector

#geting florida and kansas mask

florida_mask = state_names=="Florida"
kansas_mask = state_names=="Kansas"

#folorida or kansas evaluator
florida_or_kansas_error = florida_mask or kansas_mask #throw error because wrong operator
```

```python
#we have to use bitwise operator for element by element boolean comparison
florida_or_kansas_mask = florida_mask | kansas_mask
print("Element by element comparison for florida or kansas: \n", florida_or_kansas_mask)
```

```python
#getting data for florida or kansas
florida_or_kansas = state_co2[florida_or_kansas_mask,:]

comp116.array_to_html(florida_or_kansas, row_names=state_names[florida_or_kansas_mask],col_names=co2_categories)
```

```python
#  multiple columns using boolean selector
# coal, petrol adn natural gas i.e. the `Ttal coal`, `Total petro` and `Total nat gas` columns

coal_mask = co2_categories=="Total coal"
petrol_mask = co2_categories=="Total petro"
nat_gas_mask = co2_categories=="Total nat gas"

```

```python
# creating a mask for the col

coal_or_petrol_or_gas_mask = coal_mask|petrol_mask|nat_gas_mask
print("Multiple column mask array: ", coal_or_petrol_or_gas_mask)
```

```python
#extracting data for the mask

coal_petrol_gas = state_co2[:,coal_or_petrol_or_gas_mask]

comp116.array_to_html(coal_petrol_gas,row_names=state_names,col_names=co2_categories[coal_or_petrol_or_gas_mask])
```

```python
# can we combine row and column selectors to isolate a subset of the data?
 # we can use florida or kansas mask for row selector
 # Also for column select we can use coal_petrol_gas

#extracting data using the row and col mask

ttotal_co2 = state_co2[florida_or_kansas_mask,:][:,coal_or_petrol_or_gas_mask]

comp116.array_to_html(ttotal_co2,row_names=state_names[florida_or_kansas_mask],col_names=co2_categories[coal_or_petrol_or_gas_mask])
```

```python
#Answer:
# 1. selecting columns to display in the data table
# choosing Electrical coal, Electrical petro and Electrical nat gas
# column masking

elec_coal_mask = co2_categories == "Electrical coal"

elec_petro_mask = co2_categories == "Electrical petro"
elec_nat_gas_mask = co2_categories == "Electrical nat gas"

elec_mask = elec_coal_mask | elec_petro_mask | elec_nat_gas_mask
print("column mask: ",elec_mask)

```

```python
#Select states by comparing names *greater* than something
# selecting states >"F"
state_names_lower = np.char.lower(state_names)

row_mask = state_names_lower > "i"
print("row names that starts after F: ",row_mask)
```

```python
# 3 .Select $CO_2$ values greater than a certain value using a boolean selector 
# value greater than 1 since the elec petrol value are lesser, we will mask after extracting the data from state co2

df = state_co2[row_mask,:][:,elec_mask]
print(df.shape)

df_mask =np.any(df>1,axis=1)
print(df_mask.shape)

comp116.array_to_html(df[df_mask,:],row_names=state_names[row_mask][df_mask],col_names=co2_categories[elec_mask])


```
