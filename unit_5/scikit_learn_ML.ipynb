#### scikit learn with numpy library  
numpy works well numeric operation on array.  
To leverage the numpy's fast math capabilities.  
Scikit learn accepts numpy array for ML work flow.  
Scikit learn methods mostly expects  2D array in X(feature/sample) and 1D array in Y(target/label).

```python
## this example shows the basic linear regression 
import numpy as np
from sklearn.linear_model import LinearRegression

#create numpy array
x = np.array([[1],[2],[3],[4],[5]])  #feature 2D
y = np.array([1,2,3,4,5])         #target 1D

# print(x.ndim)
# print(y.ndim)

#create model
model = LinearRegression()

#training the model
model.fit(x,y)

#prediction
#predict expects (1 feature) 

#Single prediction (1 sample, 1 feature)
prediction = model.predict(np.array([[6]]))

# Multiple predictions (3 samples, 1 feature each)
#predictions = model.predict(np.array([[6], [7], [8]]))

print("prediction: ",prediction)

##keynote: the feature used in x must match with prediction feature
```

```python
#### divding the data as train and test sets, the training set will 20% of the dataset
x = dataset.data
y = dataset.target
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.2,random_state=1)
```

```python
## minmaxscaler()

from sklearn.preprocessing import MinMaxScaler

mm = MinMaxScaler()
mm.fit(x_train)
x_train_std = mm.transform(x_train)

## after scaling the train only we have to scale the test to avoid data leakage while regression
x_test_std_mm = MinMaxScaler().fit(x_test).transform(x_test)  # scaling the data independently of train obj mm
#we can use mm obj of train to avoid data leakage as below.
# mm.fit(x_test)
# x_test_std_mm = mm.transform(x_test)
```

```python
##Random forest classifier

#refer randomforest details in https://d.docs.live.net/6930462D258EBA80/Documents/Random%20Forest.docx
from sklearn.ensemble import RandomForestClassifier
# create model for randomforestclassifier

model = RandomForestClassifier(n_estimators=100, criterion='entropy',random_state=1)
#fiting the model
model.fit(x_train,y_train)
#prediction
y_train_pred =  model.predict(x_train)
y_test_pred = model.predict(x_test) 
print(x_test.shape)
```

```python
from sklearn.metrics import confusion_matrix
comfusion = confusion_matrix(y_true=y_test,y_pred=y_test_pred)
print(comfusion)
print(comfusion.sum())
```
