#### from dictionary to df using pandas

```python
pop_dict = {
    "Country": ["Sub-Saharan Africa", "Finland", "Gabon", "Burkina Faso", "Pakistan"],
    "1960": [227948869.0, 4429634.0, 513340.0, 4783259.0, 45954226.0],
    "1970": [293900563.0, 4606307.0, 597192.0, 5611666.0, 59290872.0],
    "1980": [388793467.0, 4779535.0, 749078.0, 6932967.0, 80624057.0],
    "1990": [516629688.0, 4986431.0, 983028.0, 9131361.0, 115414069.0],
    "2000": [671212486.0, 5176209.0, 1272935.0, 11882888.0, 154369924.0],
    "2010": [879797419.0, 5363352.0, 1711105.0, 16116845.0, 194454498.0],
    "2020": [1151302081.0, 5529543.0, 2292573.0, 21522626.0, 227196741.0]
}
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(pop_dict) #since dictionaries have the key with col name

df.head()
```

```python
#using the name of the col to get the series
df['Country']
```

```python
#we can also get by "." dot notation for non-numeric named columns
df.Country
```

```python
# you will get error if you numeric column name
df.1960
```

```python
#getting list of indices and columns

print(df.index) # this gives the range of the index like start, stop and step

print(df.columns) # list of columns
```

```python
# Series also have indices
print(df["Country"].index) #gives rangeindex
```

```python
# return as list
df.columns.to_list()
```

```python
#return numpy array
df["1960"].to_numpy()
```

```python
# we can set the new indices using pandas
df.set_index(pd.Index(["a","b","c","d","e"]))
```

```python
#but the original df is not changed since we didnt assign the last change to the df
df.head()
```

```python
#changing the country as indices
df = df.set_index("Country")
df.head()
```

```python
#resting the indices

df = df.reset_index()
df.head()
```
