#### creating sample dataframe using list

```python
vehicle =["Honda Civics","Toyota Camry","Ford f-150"]
tank_size =[12.4,15.6,26.0]
mpg = [35,32,20]
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns=["vehicle","tank_size","mpg"]) # we are creating df with columns only, not inserting the data
#df is 2 dimensional
df.head()
```

```python
df["vehicle"] = vehicle #Inserting vehicle data to vehicle column in the dataframe 
df["tank_size"] = tank_size
df["mpg"] = mpg

df.head()
```

```python
# we can add new columns from list to dataframe
#list called passenger
passenger =[5,5,3]

df["passenger"] = passenger

df.head()
```

```python
#Add new column by math
df["range"] = df["tank_size"]*df["mpg"]

df.head()
```

##### Series  
series is 1D.  
Each column in the dataframe is a series


```python
series = df["range"]
series.head()
```

```python
#returning series to list
series.to_list()
```

Not only we can create manual dataframe in python but also we can use dataframe from common spreadsheets like csv or excel files.  
Also we can export our result to csv or excel.  

```python
# reading csv file using pandas
vehicle_df = pd.read_csv("MY2023_Fuel_Consumption_Ratings.csv")

vehicle_df.head() # by default it will show the first 5 columns. 
vehicle_df.head(10) # manually entering the cunt to display.
```

```python
#Tail shows last rows
vehicle_df.tail() # by default dispalys last 5 rows
# using tail we can also find the total no.of.vehicles in the dataframe. eg: from 0 to 822 in this file
```

```python
#shape of dataframe like numpy array
vehicle_df.shape
# result shows(rows,col)
```

```python
# summary info shows col, counts and dtypes
vehicle_df.info()
```

```python
# description statistics
vehicle_df.describe()
```

```python
#save to excel
vehicle_df.to_excel("fuel_consumption.xlsx")
```

```python
#read the saved excel
excel = pd.read_excel("fuel_consumption.xlsx")

excel.head()
```
