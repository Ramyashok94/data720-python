```python
import pandas as pd

# Creating a sample DataFrame with a MultiIndex
data = {
    'Vehicle': ['Sedan', 'Sedan', 'Sedan', 'SUV', 'SUV', 'SUV', 'Truck', 'Truck'],
    'Make': ['Toyota', 'Ford', 'Honda', 'Honda', 'Acura', 'Toyota', 'Toyota', 'Ford'],
    'MPG': [33, 32, 36, 27, 24, 26, 22, 21],
}
```

```python
df = pd.DataFrame(data)

df
```

```python
#creating multiple index using set_index

multi_df = df.set_index(["Vehicle","Make"])
multi_df
```

```python
#Accessing ele from multiple index
multi_df.loc["Sedan"]
```

```python
#Accessing ele based on vehicle and make
multi_df.loc["SUV","Acura"]
```

```python
#Accessing ele with multiindex
#Syntax: DataFrame.xs(key, level=None, axis=0)
multi_df.xs("Ford",level="Make")
```

#### Sort the DataFrame (or Series) by its index and modify it directly.  
sort_index(inplace=True)  
What inplace=True Means  

inplace=True → Changes the original DataFrame.  

inplace=False (default) → Returns a new sorted DataFrame without modifying the original.  

```python
multi_df.sort_index(inplace=True)
multi_df
```

1️⃣ stack()
👉 Moves columns → into rows

(It compresses columns into a deeper row index.)

Before stack:
Rows × Columns

After stack:
More rows × Fewer columns


 2️⃣ unstack()
👉 Moves row index level → into columns

(It expands rows back into columns.)  


Before unstack:
Multi-level rows

After unstack:
Some row levels become columns

df.unstack(level=0) You can specify which level to move.

```python
unstack_df = multi_df.unstack("Make") 
unstack_df #unstacked by level "Make"
```

```python
stack_of = unstack_df.stack("Make")
stack_of
```

```python
#Aggregation using multiindex

multi_df.groupby("Make")["MPG"].mean()
```

📌 groupby() in Python (pandas)

groupby() is used to group data based on a column and perform calculations on each group.

In Python, groupby() is mainly used in the pandas library.

🔹 It is used to split data into groups, apply a function, and combine the results.

This follows the pattern:

Split → Apply → Combine

#### syntax:  
df.groupby("column_name")
df.groupby("Department")["Salary"].sum()  
👉 It groups rows by Department, then adds Salary in each group.
 
df.groupby("Department")["Salary"].agg(["sum", "mean", "max"])

```python
#groupby Vehicle
multi_df.groupby("Vehicle")["MPG"].mean()
```

```python
#resetting the multi index

multi_df.reset_index()
```
