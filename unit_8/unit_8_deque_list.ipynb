```python
from collections import deque
```

```python
my_deque = deque()
```

```python
my_deque.append(10)
my_deque.append(20)
my_deque.append(30)
my_deque.append(40)
my_deque.append(50)
my_deque
```

```python
#poping an element form deque using pop()

my_deque.pop() #it will pop the last element in the deque 
```

```python
my_deque.popleft() #removes first element and return it
```

```python
#setting maxlength to the deque, If a new element added and the length is more than the maxlength, it will delete the older element in the list

bounded = deque(maxlen=3)
```

```python
bounded.append(10)
bounded.append(20)
bounded.append(30)
bounded.append(40)
bounded.append(50)
bounded
```

```python
#differentiating list and deque bytes

import sys
import matplotlib.pyplot as plt
```

```python
list_bytes =[]
deque_bytes=[]

x = [i for i in range(1,1000,10)]

for length in range(1,1000,10):
    listed =[ i for i in range(length)]
    list_bytes.append(sys.getsizeof(listed))
    deque_bytes.append(sys.getsizeof(deque(listed)))

plt.plot(x,list_bytes,label="listed")
plt.plot(x,deque_bytes,label="deque")
plt.xlabel("Array size")
plt.ylabel("Range value")
plt.legend()
plt.show()
```
